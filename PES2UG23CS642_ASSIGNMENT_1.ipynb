EXPERIMENT 1

```python
from transformers import pipeline
```

```python
prompt = "The future of Artificial Intelligence is"
```

```python
print("BERT Text Generation")

try:
    bert_gen = pipeline("text-generation", model="bert-base-uncased")
    output = bert_gen(prompt, max_length=50)
    print(output)
except Exception as e:
    print("Error:", e)
```

```python
print("\nRoBERTa Text Generation")

try:
    roberta_gen = pipeline("text-generation", model="roberta-base")
    output = roberta_gen(prompt, max_length=50)
    print(output)
except Exception as e:
    print("Error:", e)
```

```python
print("\nBART Text Generation")

bart_gen = pipeline("text-generation", model="facebook/bart-base")
output = bart_gen(prompt, max_length=50)
print(output)
```

EXPERIMENT 2


```python
masked_sentence = "The goal of Generative AI is to [MASK] new content."
```

```python
print("BERT Fill-Mask")

bert_fill = pipeline("fill-mask", model="bert-base-uncased")
bert_results = bert_fill(masked_sentence)

for res in bert_results:
    print(res["token_str"], ":", res["score"])
```

```python
print("\nRoBERTa Fill-Mask")

# RoBERTa uses <mask> instead of [MASK]
roberta_sentence = "The goal of Generative AI is to <mask> new content."

roberta_fill = pipeline("fill-mask", model="roberta-base")
roberta_results = roberta_fill(roberta_sentence)

for res in roberta_results:
    print(res["token_str"], ":", res["score"])
```

```python
bart_sentence = "The goal of Generative AI is to <mask> new content."

print("\nBART Fill-Mask")
bart_fill = pipeline("fill-mask", model="facebook/bart-base")
bart_results = bart_fill(bart_sentence)

for res in bart_results:
    print(res["token_str"], ":", res["score"])
```

EXPERIMENT 3

```python
context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
question = "What are the risks?"
```

```python
print("BERT Question Answering")

bert_qa = pipeline("question-answering", model="bert-base-uncased")
result = bert_qa(question=question, context=context)
print(result)
```

```python
print("\nRoBERTa Question Answering")

roberta_qa = pipeline("question-answering", model="roberta-base")
result = roberta_qa(question=question, context=context)
print(result)
```

```python
print("\nBART Question Answering")

bart_qa = pipeline("question-answering", model="facebook/bart-base")
result = bart_qa(question=question, context=context)
print(result)
```
